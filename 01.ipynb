```python
import os
base = "PetImages"
for cls in ["Cat", "Dog"]:
    n = len(os.listdir(os.path.join(base, cls)))
    print(cls, n)   # should print ~12499 each
from PIL import Image
import os

def verify_images(folder):
    bad = []
    for fname in os.listdir(folder):
        path = os.path.join(folder, fname)
        try:
            with Image.open(path) as img:
                img.verify()
        except Exception:
            bad.append(path)
    return bad

for cls in ["Cat", "Dog"]:
    bad = verify_images(os.path.join(base, cls))
    print(cls, "corrupted:", len(bad))
    for p in bad:
        os.remove(p)
```
